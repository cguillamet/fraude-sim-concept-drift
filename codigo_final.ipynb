```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, recall_score, confusion_matrix
from sklearn.preprocessing import LabelEncoder
from xgboost import XGBClassifier
import warnings
from typing import List

# Configuración inicial
warnings.filterwarnings('ignore')
sns.set_style("whitegrid")
sns.despine(left=True, bottom=True)


# FUNCIONES


def calculate_variance_recall(recall_historico: List[float]) -> float:
    """
    Calcula la varianza del recall histórico: la dispersión de la serie
    alrededor de su propia media. Un valor menor indica mayor estabilidad
    del desempeño a lo largo del tiempo (menos sensible al drift).
    """
    recall_arr = np.array(recall_historico)
    return np.var(recall_arr)   # equivalente a tu cálculo manual, más directo


def ventanas_vecinas_prefijadas(ventana_actual):
    mapa = {
        30: [10, 7, 5],
        10: [7, 5],
        7: [5, 3]
    }
    return mapa.get(ventana_actual, [])


def comparar_con_baseline(df_metrics, baseline=30):
    df = df_metrics.copy()

    required_cols = ['Recall_Post_Drift', 'Varianza_Recall', 'Avg_Recovery_Time (Days)']
    for c in required_cols:
        if c not in df.columns:
            df['Score_Ventana'] = np.nan
            df['Comparacion_vs_30'] = f"Sin {c}"
            return df

    # Incluimos N_Drifts_Detectados solo como contexto 
    cols_mediana = required_cols + (['N_Drifts_Detectados'] if 'N_Drifts_Detectados' in df.columns else [])
    medianas = df.groupby('ventana')[cols_mediana].median()

    if baseline not in medianas.index:
        df['Score_Ventana'] = np.nan
        df['Comparacion_vs_30'] = "Sin baseline"
        return df

    base = medianas.loc[baseline]

    resultados = {}
    for v in medianas.index:
        if v == baseline:
            resultados[v] = (0, "Baseline")
            continue

        m = medianas.loc[v]
        score = 0
        if m['Recall_Post_Drift'] > base['Recall_Post_Drift']:
            score += 1
        if m['Varianza_Recall'] < base['Varianza_Recall']:
            score += 1
        if m['Avg_Recovery_Time (Days)'] < base['Avg_Recovery_Time (Days)']:
            score += 1

        leyenda = "Mejora adaptativa vs 30" if score >= 2 else "No mejora vs 30"
        resultados[v] = (score, leyenda)

    df['Score_Ventana'] = df['ventana'].map(lambda v: resultados[v][0])
    df['Comparacion_vs_30'] = df['ventana'].map(lambda v: resultados[v][1])
    # Mediana de drifts por ventana, como contexto del score
    if 'N_Drifts_Detectados' in df.columns:
        df['N_Drifts_Mediano'] = df['ventana'].map(lambda v: medianas.loc[v, 'N_Drifts_Detectados'])

    return df



# Rutas y archivos
ruta_archivo_parquet = '/Users/ceciliaguillametchargue/Tesis/Tesis_MGADO/simulated-data'
archivos = ['transactions_all_run1.parquet', 'transactions_all_run2.parquet','transactions_all_run3.parquet', 'transactions_all_run4.parquet',
            'transactions_all_run5.parquet', 'transactions_all_run6.parquet','transactions_all_run7.parquet', 'transactions_all_run8.parquet',
            'transactions_all_run9.parquet', 'transactions_all_run10.parquet'
            ]

ventana_inicial = [30]
ventanas_a_procesar = ventana_inicial.copy()
ventanas_procesadas = set()

df_metrics = pd.DataFrame()


# Inicio de iteración
while ventanas_a_procesar:
    ventana = ventanas_a_procesar.pop(0)

    if ventana in ventanas_procesadas:
        continue

    ventanas_procesadas.add(ventana)


    for run in archivos:
        
        # Semilla única por corrida
        run_index = archivos.index(run)
        run_seed = 42 + run_index
        
        print(f"\n--- Procesando archivo: {run} (Seed: {run_seed}) ---")
        
        # Carga y Preprocesamiento de Datos
        try:
            df = pd.read_parquet(f'{ruta_archivo_parquet}/{run}')
        except FileNotFoundError:
            print(f"Error: Archivo {run} no encontrado en la ruta. Saltando este run.")
            continue

        
        label_encoder_customer = LabelEncoder()
        label_encoder_terminal = LabelEncoder()

        df['CUSTOMER_ID'] = label_encoder_customer.fit_transform(df['CUSTOMER_ID'])
        df['TERMINAL_ID'] = label_encoder_terminal.fit_transform(df['TERMINAL_ID'])

        df['TX_DATETIME'] = pd.to_datetime(df['TX_DATETIME'])   
        df['TX_TIME'] = df['TX_DATETIME'].dt.hour               # extraigo la hora
        df['TX_DAY_WEEK'] = df['TX_DATETIME'].dt.weekday        # el día de la semana 
        df['TX_DATETIME'] = df['TX_DATETIME'].dt.normalize()    # trunca a día (medianoche)

        df = df[df['TX_FRAUD_SCENARIO'] != 1]
        df = df
        
        # Definimos las fechas clave para los periodos
        FECHA_CORTE_1 = '2018-06-30'
        FECHA_INICIO_TRANSICION_1 = '2018-06-01' # Comienzo de la convivencia 
        FECHA_CORTE_2 = '2018-09-30'
        FECHA_INICIO_TRANSICION_2 = '2018-09-01' # Comienzo de la convivencia para el siguiente cambio
        FECHA_CORTE_3 = '2018-12-31'
        FECHA_INICIO_TRANSICION_3 = '2018-12-01' # Comienzo de la convivencia para el siguiente cambio

        # Definición de los períodos:
        # Periodo inicial (hasta 2018-05-31): Solo Fraude 2 (o 0)
        condicion_periodo_1 = ((df['TX_DATETIME'] <= FECHA_INICIO_TRANSICION_1) & ((df['TX_FRAUD_SCENARIO'] == 2) | (df['TX_FRAUD_SCENARIO'] == 0)))

        # Periodo de Transición 1 (2018-06-01 a 2018-06-30): Convivencia Fraude 2 y Fraude 3 (o 0)
        condicion_transicion_1 = ((df['TX_DATETIME'] > FECHA_INICIO_TRANSICION_1) & (df['TX_DATETIME'] <= FECHA_CORTE_1) &
            ((df['TX_FRAUD_SCENARIO'] == 2) | (df['TX_FRAUD_SCENARIO'] == 3) | (df['TX_FRAUD_SCENARIO'] == 0)))

        # Periodo intermedio (2018-07-01 a 2018-08-31): Solo Fraude 3 (o 0)
        condicion_periodo_2 = ((df['TX_DATETIME'] > FECHA_CORTE_1) & (df['TX_DATETIME'] <= FECHA_INICIO_TRANSICION_2) &
            ((df['TX_FRAUD_SCENARIO'] == 3) | (df['TX_FRAUD_SCENARIO'] == 0)))

        # Periodo de Transición 2 (2018-09-01 a 2018-09-30): Convivencia Fraude 3 y Fraude 2 (o 0)
        condicion_transicion_2 = ((df['TX_DATETIME'] > FECHA_INICIO_TRANSICION_2) & (df['TX_DATETIME'] <= FECHA_CORTE_2) &
            ((df['TX_FRAUD_SCENARIO'] == 3) | (df['TX_FRAUD_SCENARIO'] == 2) | (df['TX_FRAUD_SCENARIO'] == 0)))

        condicion_periodo_3 = ((df['TX_DATETIME'] > FECHA_CORTE_2) & (df['TX_DATETIME'] <= FECHA_INICIO_TRANSICION_3) &
            ((df['TX_FRAUD_SCENARIO'] == 2) | (df['TX_FRAUD_SCENARIO'] == 0)))

        condicion_transicion_3 = ((df['TX_DATETIME'] > FECHA_INICIO_TRANSICION_3) & (df['TX_DATETIME'] <= FECHA_CORTE_3) &
            ((df['TX_FRAUD_SCENARIO'] == 3) | (df['TX_FRAUD_SCENARIO'] == 2) | (df['TX_FRAUD_SCENARIO'] == 0)))

        condicion_periodo_4 = ((df['TX_DATETIME'] > FECHA_CORTE_3) & ((df['TX_FRAUD_SCENARIO'] == 3) | (df['TX_FRAUD_SCENARIO'] == 0)))

        # Combinamos todas las condiciones 
        condicion_final = (
            condicion_periodo_1 |
            condicion_transicion_1 |
            condicion_periodo_2 |
            condicion_transicion_2 |
            condicion_periodo_3 |
            condicion_transicion_3 |
            condicion_periodo_4
        )

        # Aplicamos el filtro al DataFrame
        df = df[condicion_final]

        # Ordenar y crear features de ventana móvil (rolling window)
        df = df.sort_values(by=["CUSTOMER_ID", "TX_DATETIME"])
        daily_spending = df.groupby(["CUSTOMER_ID", "TX_DATETIME"])["TX_AMOUNT"].agg(
            TX_AMOUNT_DAILY="sum", TX_CNT_DAILY="count"
        ).reset_index()

        for window in [3, 7, 30]:
            daily_spending[f"TX_AMOUNT_{window}D"] = (
                daily_spending.groupby("CUSTOMER_ID")["TX_AMOUNT_DAILY"]
                .rolling(window=window, min_periods=1).sum().reset_index(0, drop=True)
            )
            daily_spending[f"MOV_AVG_TX_AMOUNT_{window}D"] = (
                daily_spending.groupby("CUSTOMER_ID")["TX_AMOUNT_DAILY"]
                .rolling(window=window, min_periods=1).mean().reset_index(0, drop=True)
            )
            daily_spending[f"TX_CNT_{window}D"] = (
                daily_spending.groupby("CUSTOMER_ID")["TX_CNT_DAILY"]
                .rolling(window=window, min_periods=1).sum().reset_index(0, drop=True)
            )
            daily_spending[f"TX_STD_AMOUNT_{window}D"] = (
                daily_spending.groupby("CUSTOMER_ID")["TX_AMOUNT_DAILY"]
                .rolling(window=window, min_periods=1).std().reset_index(0, drop=True)
            )
        daily_spending["VAR_TX_AMOUNT_1D"] = daily_spending.groupby("CUSTOMER_ID")["TX_AMOUNT_DAILY"].diff()
        df = df.merge(daily_spending, on=["CUSTOMER_ID", "TX_DATETIME"], how="left")

        # Modelado de los datos con ventana de 20 días
        accuracy_list_eval = []
        recall_list_eval = []
        
        unique_dates = sorted(df['TX_DATETIME'].unique())
        drifts_detectados = []          # lista de dicts, uno por cada drift
        en_periodo_refractario = False  # True justo después de un drift, hasta recuperarse
        dias_desde_drift = 0
        DIAS_REFRACTARIO = 30            # ventana mínima entre drifts consecutivos
        baseline_actual = np.nan

        
        for i in range(0, len(unique_dates) - ventana):
            contador_inicial = i
            contador_final = ventana + i

            df_p = df[df['TX_DATETIME'].isin(unique_dates[contador_inicial:contador_final])].copy()

            # Dividir en entrenamiento y prueba
            df_train = df_p[df_p['TX_DATETIME'].isin(unique_dates[contador_inicial:contador_final-1])]
            df_prueba = df_p[df_p['TX_DATETIME'] == unique_dates[contador_final-1]]
            
            # Submuestreo balanceado. Usamos la semilla única del run.
            try:
                df_ent = df_train[df_train['TX_FRAUD'] == 0].sample(n=df_train[df_train['TX_FRAUD'] == 1].shape[0], random_state=run_seed)
            except ValueError:
                continue 

            df_ent = pd.concat([df_ent, df_train[df_train['TX_FRAUD'] == 1]])
            
            features_to_drop = ['TX_DATETIME', 'TX_FRAUD_SCENARIO', 'TX_FRAUD', 
                    'TRANSACTION_ID', 'TX_DATE', 'TX_TIME_SECONDS', 'TX_TIME_DAYS']

            columns_to_keep = [c for c in df_ent.columns if c not in features_to_drop]

            X_train = df_ent[columns_to_keep]
            y_train = df_ent['TX_FRAUD']
            X_test = df_prueba[columns_to_keep]
            y_test = df_prueba['TX_FRAUD']

            # Ajuste del modelo
            # Usamos la semilla única del run para XGBoost
            model_f23 = XGBClassifier(
                n_estimators=100, max_depth=6, learning_rate=0.1, 
                objective='binary:logistic', random_state=run_seed, eval_metric='logloss'
            )
            model_f23.fit(X_train, y_train)

            # Predicción y cálculo de métricas
            y_pred_eval = model_f23.predict(X_test)
            accuracy_eval = accuracy_score(y_test, y_pred_eval)
            recall_eval = recall_score(y_test, y_pred_eval)
            tn, fp, fn, tp = confusion_matrix(y_test, y_pred_eval, labels=[0, 1]).ravel().tolist()
            
            accuracy_list_eval.append(accuracy_eval)
            recall_list_eval.append(recall_eval)


            
            # SISTEMA DE ALARMAS
            

            UMBRAL_RECALL = 0.70          # recall mínimo permitido
            UMBRAL_RELATIVO = 0.25        # caída > 15% del baseline dispara alarma
            PERSISTENCIA_DIAS = 3         # caída debe durar 3 días seguidos

            # --- Alarma instantánea ---
            if recall_eval < UMBRAL_RECALL:
                print(f"ALERTA: Recall bajo ({recall_eval:.3f}) en fecha = {unique_dates[contador_final-1]}")

            # --- Alarma por persistencia ---
            if len(recall_list_eval) >= PERSISTENCIA_DIAS:
                ultimos = recall_list_eval[-PERSISTENCIA_DIAS:]
                if all(r < UMBRAL_RECALL for r in ultimos):
                    print(f"ALERTA PERSISTENTE: Recall < {UMBRAL_RECALL} por {PERSISTENCIA_DIAS} días consecutivos.")
                    print(f"Últimos valores = {ultimos}")

            # --- Alarma por cambio estructural (indicación de nuevo caso de fraude) ---
            # --- Manejo del período refractario ---
            if en_periodo_refractario:
                dias_desde_drift += 1
                # Salimos del refractario si el recall se recuperó O pasaron suficientes días
                if recall_eval >= baseline_actual or dias_desde_drift >= DIAS_REFRACTARIO:
                    en_periodo_refractario = False
                    dias_desde_drift = 0

            # Antes de la detección, calculá el recall suavizado
            VENTANA_SUAVIZADO = 3
            if len(recall_list_eval) >= VENTANA_SUAVIZADO:
                recall_suavizado = np.median(recall_list_eval[-VENTANA_SUAVIZADO:])
            else:
                recall_suavizado = recall_eval

            fecha_actual = unique_dates[contador_final-1]
            if pd.Timestamp('2018-09-15') <= pd.Timestamp(fecha_actual) <= pd.Timestamp('2018-10-15'):
                bl = np.median(recall_list_eval[:-1]) if len(recall_list_eval) > 1 else np.nan
                caida = (bl - recall_suavizado) / bl if bl and bl > 0 else np.nan
                print(f"{pd.Timestamp(fecha_actual).date()} | crudo={recall_eval:.2f} | "
                    f"suav={recall_suavizado:.2f} | baseline={bl:.2f} | "
                    f"caída={caida:.2f} | refractario={en_periodo_refractario}")

            # --- Detección de drift 
            if len(recall_list_eval) > 1 and not en_periodo_refractario:
                baseline_actual = np.median(recall_list_eval[:-1])
                if baseline_actual > 0:
                    caida_relativa = (baseline_actual - recall_suavizado) / baseline_actual

                    if caida_relativa > UMBRAL_RELATIVO:
                        # Baseline pre-drift: mediana de los K días previos
                        K = 5
                        if len(recall_list_eval) >= K + 1:
                            recall_pre_drift = recall_list_eval[-K-1:-1]
                        else:
                            recall_pre_drift = recall_list_eval[:-1]

                        drifts_detectados.append({
                            'fecha': unique_dates[contador_final-1],
                            'indice': contador_final - 1,
                            'indice_local': len(recall_list_eval) - 1,  
                            'baseline_pre_drift': np.median(recall_pre_drift),
                            'recall_en_drift': recall_eval,
                            'caida_relativa': caida_relativa,
                        })

                        # Activamos refractario para no re-disparar inmediatamente
                        en_periodo_refractario = True
                        dias_desde_drift = 0

                        # Disparo de corridas adaptativas
                        ventanas_candidatas = ventanas_vecinas_prefijadas(ventana)
                        for v in ventanas_candidatas:
                            if v not in ventanas_procesadas and v not in ventanas_a_procesar:
                                print(f"→ Programando corrida adaptativa con ventana = {v}")
                                ventanas_a_procesar.append(v)
        
        # Cálculo de Métricas de Penalización (Varianza y Max Drop)
        if not accuracy_list_eval:
            print(f"Advertencia: No se pudo realizar la evaluación para {run}.")
            continue

        df_acc = pd.DataFrame({
            'Accuracy_eval': accuracy_list_eval,
            'Recall_eval': recall_list_eval})

        # Calculamos el Recall Promedio para el Run
        current_mean_recall = np.mean(df_acc['Recall_eval'])
        
        
        # CÁLCULO DE LA VARIANZA DEL RECALL (Target = Promedio)
        
        
        # ⬅️ Llamamos a la nueva función sin target
        varianza_recall = calculate_variance_recall(df_acc['Recall_eval'].tolist())
        print(f"Varianza del Recall (vs promedio {current_mean_recall:.4f}) para {run}: {varianza_recall:.6f}")
        
        
        # Cálculo del Tiempo Promedio de Recuperación (Avg_Recovery_Time)
        
        
        recall_series = df_acc['Recall_eval']
        baseline_recall = recall_series.median()
        # Se asegura que el umbral sea un valor positivo o significativo
        drop_threshold = max(0.01, baseline_recall - 0.05) 
        recovery_times = []
        
        for i in range(len(recall_series)):
            current_recall = recall_series.iloc[i]
            if current_recall < drop_threshold:
                days_to_recover = 0
                is_recovered = False
                for j in range(i + 1, len(recall_series)):
                    days_to_recover += 1
                    if recall_series.iloc[j] >= baseline_recall:
                        recovery_times.append(days_to_recover)
                        is_recovered = True
                        break
                if not is_recovered:
                    recovery_times.append(np.nan) 

        if recovery_times and not all(np.isnan(recovery_times)):
            avg_recovery_time = np.nanmean(recovery_times)
        else:
            avg_recovery_time = np.nan

        
        # CÁLCULO DE LA PROFUNDIDAD MÁXIMA DE CAÍDA (MAX DROP MAGNITUDE)
        

        drop_depth_series = baseline_recall - recall_series
        significant_drops = drop_depth_series[recall_series < drop_threshold]
        
        if not significant_drops.empty:
            max_drop_magnitude = significant_drops.max()
        else:
            max_drop_magnitude = 0.0
        
        print(f"Profundidad Máxima de Caída para {run}: {max_drop_magnitude:.4f}")

        # =========================================================================

        K = 5
        metricas_por_drift = []

        for d in drifts_detectados:
            start = d['indice_local']
            end = min(start + K, len(recall_list_eval))
            recall_post = recall_list_eval[start:end]

            # Recall mediano en la ventana post-drift
            recall_post_drift = np.median(recall_post) if recall_post else np.nan

            # Tiempo de recuperación: días hasta volver al baseline pre-drift
            tiempo_recuperacion = np.nan
            for offset, r in enumerate(recall_list_eval[start:], start=0):
                if r >= d['baseline_pre_drift']:
                    tiempo_recuperacion = offset
                    break

            metricas_por_drift.append({
                'fecha': d['fecha'],
                'baseline_pre_drift': d['baseline_pre_drift'],
                'recall_post_drift': recall_post_drift,
                'caida_relativa': d['caida_relativa'],
                'tiempo_recuperacion': tiempo_recuperacion,
            })

        
        # VALIDACIÓN: drifts detectados vs transiciones conocidas
        
        fechas_corte = [pd.Timestamp('2018-06-30'),
                        pd.Timestamp('2018-09-30'),
                        pd.Timestamp('2018-12-31')]
        TOLERANCIA_DIAS = 30

        print(f"\n  Drifts detectados: {len(metricas_por_drift)}")
        for m in metricas_por_drift:
            fecha = pd.Timestamp(m['fecha'])
            dist_min = min(abs((fecha - fc).days) for fc in fechas_corte)
            acierto = "✓ cerca de transición" if dist_min <= TOLERANCIA_DIAS else "✗ espurio (lejos de toda transición)"
            print(f"    {fecha.date()} | caída={m['caida_relativa']:.2f} | "
                  f"a {dist_min}d del corte más cercano | {acierto}")

        capturadas = sum(
            any(abs((pd.Timestamp(m['fecha']) - fc).days) <= TOLERANCIA_DIAS
                for m in metricas_por_drift)
            for fc in fechas_corte
        )
        print(f"  Transiciones reales capturadas: {capturadas}/3")

        # Métricas agregadas para la fila de df_metrics
        n_drifts = len(metricas_por_drift)
        if n_drifts > 0:
            recall_post_drift_promedio = np.nanmean([m['recall_post_drift'] for m in metricas_por_drift])
            tiempo_recup_promedio = np.nanmean([m['tiempo_recuperacion'] for m in metricas_por_drift])
            caida_promedio = np.nanmean([m['caida_relativa'] for m in metricas_por_drift])
        else:
            recall_post_drift_promedio = np.nan
            tiempo_recup_promedio = np.nan
            caida_promedio = np.nan



        # Almacenamiento de la métrica final 
        last_accuracy = accuracy_list_eval[-1]
        last_recall = recall_list_eval[-1]
        
        new_row = pd.DataFrame([{
            'ventana': ventana,
            'run': run,
            'Accuracy': last_accuracy,
            'Recall': last_recall,
            'Recall_Post_Drift': recall_post_drift_promedio,
            'N_Drifts_Detectados': n_drifts,
            'Tiempo_Recup_Promedio': tiempo_recup_promedio,
            'Caida_Promedio': caida_promedio,
            'Fecha_Drift': [m['fecha'] for m in metricas_por_drift],  # lista de fechas, o el detalle completo
            'Detalle_Drifts': metricas_por_drift,  # si querés inspeccionar drift por drift después
            'Varianza_Recall': varianza_recall,
            'Avg_Recovery_Time (Days)': avg_recovery_time,
            'Max_Drop_Magnitude': max_drop_magnitude,
            'Motivo_Corrida': ('Base' if ventana == 30 else 'Adaptativa')
        }])




        df_metrics = pd.concat([df_metrics, new_row], ignore_index=True)
        

        # Generación de Gráficos 
        plt.figure(figsize=(20, 10))
        sns_plot = sns.lineplot(x=df_acc.index, y=df_acc['Recall_eval'], label="Recall", linewidth=3)
        sns_plot.axhline(baseline_recall, color='green', linestyle='--', linewidth=1, label=f'Baseline Recall ({baseline_recall:.2f})')

        sns_plot.set(xticks=list(range(0, len(df_acc.index)+1, 10)), yticks=np.arange(0, 1.1, 0.1))
        ax = plt.gca()


        # LEYENDA DE SELECCIÓN DE VENTANA ÓPTIMA

        df_sel = comparar_con_baseline(
                df_metrics[df_metrics['run'] == run])

        df_sel_valid = df_sel.dropna(subset=['Score_Ventana'])

        if not df_sel_valid.empty:
            ventana_optima = df_sel_valid.loc[
                df_sel_valid['Score_Ventana'].idxmax(), 'ventana'
            ]
        else:
            ventana_optima = 30

        fila_optima = df_sel[df_sel['ventana'] == ventana_optima].iloc[0]
        n_drifts_txt = ""
        if 'N_Drifts_Mediano' in df_sel.columns and not pd.isna(fila_optima.get('N_Drifts_Mediano')):
            n_drifts_txt = f"\nDrifts detectados (mediana): {fila_optima['N_Drifts_Mediano']:.0f}"

        texto = (
            f"Ventana seleccionada: {ventana_optima} días\n"
            f"{fila_optima['Comparacion_vs_30']}"
            f"{n_drifts_txt}"
        )

        ax.text(
            0.02, 0.02,
            texto,
            transform=ax.transAxes,
            fontsize=15,
            verticalalignment="bottom",
            bbox=dict(boxstyle="round,pad=0.4", fc="whitesmoke", ec="gray")
        )


        ax.tick_params(axis='x', labelsize=24, labelrotation=0)
        ax.tick_params(axis='y', labelsize=24, labelrotation=0)
        plt.setp(ax.get_xticklabels(), fontname='Gill Sans', ha='center')
        plt.setp(ax.get_yticklabels(), fontname='Gill Sans')

        sns_plot.set_xlabel("Número de Día (Ventana)", fontsize=20, labelpad=15, fontname='Gill Sans', fontweight='bold')
        sns_plot.set_ylabel("Valor de la Métrica", fontsize=20, labelpad=15, fontname='Gill Sans', fontweight='bold')
        sns_plot.legend(loc='lower right', fontsize=15)

        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.spines['bottom'].set_visible(False)
        ax.spines['left'].set_visible(False)

        plt.tight_layout()
        plt.show()

    if not df_metrics.empty:
        print("\n\n" + "="*50)
        print("Métricas Consolidadas para todos los Runs:")
        print("="*50)
        # Formato actualizado para la Varianza
        print(df_metrics.to_string(
        formatters={
            'Accuracy': lambda x: "%.4f" % x,
            'Recall': lambda x: "%.4f" % x,
            # Renombrado y ajuste de formato
            'Varianza_Recall': lambda x: "%.6f" % x,  
            'Avg_Recovery_Time (Days)': lambda x: "%.2f" % x,
            'Max_Drop_Magnitude': lambda x: "%.4f" % x
        }))
    else:
        print("\nNo se pudieron generar métricas para ningún run.")


# RESULTADO FINAL CONSOLIDADO
if not df_metrics.empty:
    print("\n\n" + "="*50)
    print("Métricas Consolidadas para todos los Runs:")
    print("="*50)
    # Formato actualizado para la Varianza
    print(df_metrics.to_string(
    formatters={
        'Accuracy': lambda x: "%.4f" % x,
        'Recall': lambda x: "%.4f" % x,
        # Renombrado y ajuste de formato
        'Varianza_Recall': lambda x: "%.6f" % x,  
        'Avg_Recovery_Time (Days)': lambda x: "%.2f" % x,
        'Max_Drop_Magnitude': lambda x: "%.4f" % x
    }))
else:
    print("\nNo se pudieron generar métricas para ningún run.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.inspection import permutation_importance
from xgboost import XGBClassifier
import warnings

warnings.filterwarnings('ignore')
sns.set_style("whitegrid")


# CONFIGURACIÓN

RUTA_PARQUET = '/Users/ceciliaguillametchargue/Tesis/Tesis_MGADO/simulated-data/'

ARCHIVO = 'transactions_all_run1.parquet'          # cambiar o iterar sobre varios
SEED = 42

FEATURES_TO_DROP = ['TX_DATETIME', 'TX_FRAUD_SCENARIO', 'TX_FRAUD',
                    'TRANSACTION_ID', 'TX_DATE', 'TX_TIME_SECONDS', 'TX_TIME_DAYS']

# Regímenes PUROS (se evitan junio, septiembre y diciembre = meses de transición)
REGIMENES = {
    'Escenario 2 (pre)':  ('2018-01-01', '2018-05-31'),
    'Escenario 3 (x5)':   ('2018-07-01', '2018-08-31'),
    'Escenario 2 (post)': ('2018-10-01', '2018-11-30'),
}

FAMILIA_MONTO = ['TX_AMOUNT', 'TX_AMOUNT_DAILY',
                 'TX_AMOUNT_3D', 'TX_AMOUNT_7D', 'TX_AMOUNT_30D',
                 'MOV_AVG_TX_AMOUNT_3D', 'MOV_AVG_TX_AMOUNT_7D', 'MOV_AVG_TX_AMOUNT_30D',
                 'TX_STD_AMOUNT_3D', 'TX_STD_AMOUNT_7D', 'TX_STD_AMOUNT_30D',
                 'VAR_TX_AMOUNT_1D']



# PREPROCESAMIENTO 

def cargar_y_preprocesar(ruta, archivo):
    df = pd.read_parquet(f'{ruta}/{archivo}')

    # Label encoding
    df['CUSTOMER_ID'] = LabelEncoder().fit_transform(df['CUSTOMER_ID'])
    df['TERMINAL_ID'] = LabelEncoder().fit_transform(df['TERMINAL_ID'])

    # Fechas y features temporales 
    df['TX_DATETIME'] = pd.to_datetime(df['TX_DATETIME'])
    df['TX_TIME'] = df['TX_DATETIME'].dt.hour
    df['TX_DAY_WEEK'] = df['TX_DATETIME'].dt.weekday
    df['TX_DATETIME'] = df['TX_DATETIME'].dt.normalize()

    df = df[df['TX_FRAUD_SCENARIO'] != 1]

    # Filtrado por períodos/transiciones
    FC1, FT1 = '2018-06-30', '2018-06-01'
    FC2, FT2 = '2018-09-30', '2018-09-01'
    FC3, FT3 = '2018-12-31', '2018-12-01'

    c_p1 = (df['TX_DATETIME'] <= FT1) & (df['TX_FRAUD_SCENARIO'].isin([2, 0]))
    c_t1 = (df['TX_DATETIME'] > FT1) & (df['TX_DATETIME'] <= FC1) & (df['TX_FRAUD_SCENARIO'].isin([2, 3, 0]))
    c_p2 = (df['TX_DATETIME'] > FC1) & (df['TX_DATETIME'] <= FT2) & (df['TX_FRAUD_SCENARIO'].isin([3, 0]))
    c_t2 = (df['TX_DATETIME'] > FT2) & (df['TX_DATETIME'] <= FC2) & (df['TX_FRAUD_SCENARIO'].isin([3, 2, 0]))
    c_p3 = (df['TX_DATETIME'] > FC2) & (df['TX_DATETIME'] <= FT3) & (df['TX_FRAUD_SCENARIO'].isin([2, 0]))
    c_t3 = (df['TX_DATETIME'] > FT3) & (df['TX_DATETIME'] <= FC3) & (df['TX_FRAUD_SCENARIO'].isin([3, 2, 0]))
    c_p4 = (df['TX_DATETIME'] > FC3) & (df['TX_FRAUD_SCENARIO'].isin([3, 0]))

    df = df[c_p1 | c_t1 | c_p2 | c_t2 | c_p3 | c_t3 | c_p4]

    # Varibales ventana movil
    df = df.sort_values(by=["CUSTOMER_ID", "TX_DATETIME"])
    daily = df.groupby(["CUSTOMER_ID", "TX_DATETIME"])["TX_AMOUNT"].agg(
        TX_AMOUNT_DAILY="sum", TX_CNT_DAILY="count").reset_index()

    for w in [3, 7, 30]:
        g = daily.groupby("CUSTOMER_ID")["TX_AMOUNT_DAILY"]
        daily[f"TX_AMOUNT_{w}D"] = g.rolling(w, min_periods=1).sum().reset_index(0, drop=True)
        daily[f"MOV_AVG_TX_AMOUNT_{w}D"] = g.rolling(w, min_periods=1).mean().reset_index(0, drop=True)
        daily[f"TX_CNT_{w}D"] = daily.groupby("CUSTOMER_ID")["TX_CNT_DAILY"].rolling(w, min_periods=1).sum().reset_index(0, drop=True)
        daily[f"TX_STD_AMOUNT_{w}D"] = g.rolling(w, min_periods=1).std().reset_index(0, drop=True)
    daily["VAR_TX_AMOUNT_1D"] = daily.groupby("CUSTOMER_ID")["TX_AMOUNT_DAILY"].diff()

    df = df.merge(daily, on=["CUSTOMER_ID", "TX_DATETIME"], how="left")
    return df


# =========================================================================
# FUNCIONES DE ANÁLISIS
# =========================================================================
def preparar_X_y(df_periodo):
    frauds = df_periodo[df_periodo['TX_FRAUD'] == 1]
    n_fraud = len(frauds)
    if n_fraud == 0:
        return None, None
    n_no = min(n_fraud, (df_periodo['TX_FRAUD'] == 0).sum())
    no_fraud = df_periodo[df_periodo['TX_FRAUD'] == 0].sample(n=n_no, random_state=SEED)
    df_bal = pd.concat([frauds, no_fraud])
    cols = [c for c in df_bal.columns if c not in FEATURES_TO_DROP]
    return df_bal[cols], df_bal['TX_FRAUD']


def entrenar(X, y):
    m = XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.1,
                      objective='binary:logistic', random_state=SEED, eval_metric='logloss')
    m.fit(X, y)
    return m


def imp_nativa(model, cols):
    s = pd.Series(model.feature_importances_, index=cols)
    return (s / s.sum()).sort_values(ascending=False)


def imp_perm(model, X, y, n_repeats=10):
    r = permutation_importance(model, X, y, scoring='recall',
                               n_repeats=n_repeats, random_state=SEED, n_jobs=-1)
    s = pd.Series(r.importances_mean, index=X.columns).clip(lower=0)
    tot = s.sum()
    return (s / tot if tot > 0 else s).sort_values(ascending=False)


def peso_monto(serie):
    presentes = [f for f in FAMILIA_MONTO if f in serie.index]
    return serie[presentes].sum()


# --------------------------------------------------------

print(f"Cargando y preprocesando {ARCHIVO}...")
df = cargar_y_preprocesar(RUTA_PARQUET, ARCHIVO)
print(f"Filas tras preprocesamiento: {len(df)}")


print("\n" + "=" * 70)
print("1. RANKING GLOBAL")
print("=" * 70)
X_g, y_g = preparar_X_y(df)
model_g = entrenar(X_g, y_g)
ing = imp_nativa(model_g, X_g.columns)
ipg = imp_perm(model_g, X_g, y_g)
print("\nTop 15 (nativa por ganancia):")
print(ing.head(15).to_string())
print("\nTop 15 (permutación sobre recall):")
print(ipg.head(15).to_string())
print(f"\nPeso familia MONTO -> nativa: {peso_monto(ing):.3f} | permutación: {peso_monto(ipg):.3f}")


print("\n" + "=" * 70)
print("2. IMPORTANCIA POR RÉGIMEN")
print("=" * 70)
res_reg = {}
peso_reg = {}
for nombre, (ini, fin) in REGIMENES.items():
    dfr = df[(df['TX_DATETIME'] >= ini) & (df['TX_DATETIME'] <= fin)]
    Xr, yr = preparar_X_y(dfr)
    if Xr is None or yr.sum() < 5:
        print(f"\n[{nombre}] insuficientes fraudes, se omite.")
        continue
    mr = entrenar(Xr, yr)
    inr = imp_nativa(mr, Xr.columns)
    ipr = imp_perm(mr, Xr, yr)
    res_reg[nombre] = inr
    peso_reg[nombre] = {'nativa': peso_monto(inr), 'permutacion': peso_monto(ipr)}
    print(f"\n--- {nombre} --- ({int(yr.sum())} fraudes)")
    for var, val in inr.head(8).items():
        marca = "  <-- MONTO" if var in FAMILIA_MONTO else ""
        print(f"    {var:<28} {val:.3f}{marca}")
    print(f"    Peso MONTO -> nativa: {peso_reg[nombre]['nativa']:.3f} | perm: {peso_reg[nombre]['permutacion']:.3f}")

# Resumen ---
print("\n" + "=" * 70)
print("3. PESO DE LA FAMILIA DE MONTO POR RÉGIMEN")
print("=" * 70)
resumen = pd.DataFrame(peso_reg).T
print(resumen.to_string(float_format=lambda x: f"{x:.3f}"))

# Graficos  ---
plt.figure(figsize=(10, 8))
top = ipg.head(15)[::-1]
colores = ['#D85A30' if v in FAMILIA_MONTO else '#185FA5' for v in top.index]
plt.barh(range(len(top)), top.values, color=colores)
plt.yticks(range(len(top)), top.index)
plt.xlabel("Importancia por permutación (normalizada)")
plt.title("Ranking global de importancia\n(naranja = familia de monto)")
plt.tight_layout()
plt.savefig("fig_importance_global.png", dpi=300, bbox_inches="tight")
plt.close()

if peso_reg:
    fig, ax = plt.subplots(figsize=(10, 6))
    nombres = list(peso_reg.keys())
    nat = [peso_reg[n]['nativa'] for n in nombres]
    per = [peso_reg[n]['permutacion'] for n in nombres]
    x = np.arange(len(nombres)); w = 0.35
    ax.bar(x - w/2, nat, w, label='Nativa', color='#1D9E75')
    ax.bar(x + w/2, per, w, label='Permutación', color='#7A52A3')
    ax.set_xticks(x); ax.set_xticklabels(nombres, fontsize=9)
    ax.set_ylabel("Peso relativo familia de monto")
    ax.set_title("Reorganización del peso de las variables de monto entre regímenes")
    ax.legend()
    plt.tight_layout()
    plt.savefig("fig_importance_por_regimen.png", dpi=300, bbox_inches="tight")
    plt.close()

print("\nFiguras guardadas.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import recall_score
from sklearn.preprocessing import LabelEncoder
from xgboost import XGBClassifier
import warnings

warnings.filterwarnings('ignore')


# CONFIGURACIÓN

RUTA = '/Users/ceciliaguillametchargue/Tesis/Tesis_MGADO/simulated-data'
ARCHIVOS = ['transactions_all_run1.parquet', 'transactions_all_run2.parquet','transactions_all_run3.parquet', 'transactions_all_run4.parquet',
            'transactions_all_run5.parquet', 'transactions_all_run6.parquet','transactions_all_run7.parquet', 'transactions_all_run8.parquet',
            'transactions_all_run9.parquet', 'transactions_all_run10.parquet'
            ]

RUN = 'transactions_all_run3.parquet'
RUN_SEED = 42 + ARCHIVOS.index(RUN) 
VENTANA = 30

UMBRAL_RELATIVO = 0.25
VENTANA_SUAVIZADO = 3
DIAS_REFRACTARIO = 30


COLOR_RECALL = '#185FA5'      
COLOR_BASELINE = '#7A52A3'    
COLOR_DRIFT = '#D85A30'       

FECHAS_CORTE = {
    'jun': pd.Timestamp('2018-06-30'),
    'sep': pd.Timestamp('2018-09-30'),
    'dic': pd.Timestamp('2018-12-31'),
}


# 1. CARGA Y PREPROCESAMIENTO

df = pd.read_parquet(f'{RUTA}/{RUN}')

le_customer = LabelEncoder()
le_terminal = LabelEncoder()
df['CUSTOMER_ID'] = le_customer.fit_transform(df['CUSTOMER_ID'])
df['TERMINAL_ID'] = le_terminal.fit_transform(df['TERMINAL_ID'])

df['TX_DATETIME'] = pd.to_datetime(df['TX_DATETIME'])
df['TX_TIME'] = df['TX_DATETIME'].dt.hour
df['TX_DAY_WEEK'] = df['TX_DATETIME'].dt.weekday
df['TX_DATETIME'] = df['TX_DATETIME'].dt.normalize()

df = df[df['TX_FRAUD_SCENARIO'] != 1]

FECHA_CORTE_1, FECHA_INICIO_TRANSICION_1 = '2018-06-30', '2018-06-01'
FECHA_CORTE_2, FECHA_INICIO_TRANSICION_2 = '2018-09-30', '2018-09-01'
FECHA_CORTE_3, FECHA_INICIO_TRANSICION_3 = '2018-12-31', '2018-12-01'

condicion_periodo_1 = ((df['TX_DATETIME'] <= FECHA_INICIO_TRANSICION_1) &
                        ((df['TX_FRAUD_SCENARIO'] == 2) | (df['TX_FRAUD_SCENARIO'] == 0)))
condicion_transicion_1 = ((df['TX_DATETIME'] > FECHA_INICIO_TRANSICION_1) & (df['TX_DATETIME'] <= FECHA_CORTE_1) &
                           ((df['TX_FRAUD_SCENARIO'] == 2) | (df['TX_FRAUD_SCENARIO'] == 3) | (df['TX_FRAUD_SCENARIO'] == 0)))
condicion_periodo_2 = ((df['TX_DATETIME'] > FECHA_CORTE_1) & (df['TX_DATETIME'] <= FECHA_INICIO_TRANSICION_2) &
                        ((df['TX_FRAUD_SCENARIO'] == 3) | (df['TX_FRAUD_SCENARIO'] == 0)))
condicion_transicion_2 = ((df['TX_DATETIME'] > FECHA_INICIO_TRANSICION_2) & (df['TX_DATETIME'] <= FECHA_CORTE_2) &
                           ((df['TX_FRAUD_SCENARIO'] == 3) | (df['TX_FRAUD_SCENARIO'] == 2) | (df['TX_FRAUD_SCENARIO'] == 0)))
condicion_periodo_3 = ((df['TX_DATETIME'] > FECHA_CORTE_2) & (df['TX_DATETIME'] <= FECHA_INICIO_TRANSICION_3) &
                        ((df['TX_FRAUD_SCENARIO'] == 2) | (df['TX_FRAUD_SCENARIO'] == 0)))
condicion_transicion_3 = ((df['TX_DATETIME'] > FECHA_INICIO_TRANSICION_3) & (df['TX_DATETIME'] <= FECHA_CORTE_3) &
                           ((df['TX_FRAUD_SCENARIO'] == 3) | (df['TX_FRAUD_SCENARIO'] == 2) | (df['TX_FRAUD_SCENARIO'] == 0)))
condicion_periodo_4 = ((df['TX_DATETIME'] > FECHA_CORTE_3) & ((df['TX_FRAUD_SCENARIO'] == 3) | (df['TX_FRAUD_SCENARIO'] == 0)))

df = df[condicion_periodo_1 | condicion_transicion_1 | condicion_periodo_2 |
        condicion_transicion_2 | condicion_periodo_3 | condicion_transicion_3 | condicion_periodo_4]

df = df.sort_values(by=["CUSTOMER_ID", "TX_DATETIME"])
daily_spending = df.groupby(["CUSTOMER_ID", "TX_DATETIME"])["TX_AMOUNT"].agg(
    TX_AMOUNT_DAILY="sum", TX_CNT_DAILY="count"
).reset_index()

for window in [3, 7, 30]:
    daily_spending[f"TX_AMOUNT_{window}D"] = (
        daily_spending.groupby("CUSTOMER_ID")["TX_AMOUNT_DAILY"]
        .rolling(window=window, min_periods=1).sum().reset_index(0, drop=True))
    daily_spending[f"MOV_AVG_TX_AMOUNT_{window}D"] = (
        daily_spending.groupby("CUSTOMER_ID")["TX_AMOUNT_DAILY"]
        .rolling(window=window, min_periods=1).mean().reset_index(0, drop=True))
    daily_spending[f"TX_CNT_{window}D"] = (
        daily_spending.groupby("CUSTOMER_ID")["TX_CNT_DAILY"]
        .rolling(window=window, min_periods=1).sum().reset_index(0, drop=True))
    daily_spending[f"TX_STD_AMOUNT_{window}D"] = (
        daily_spending.groupby("CUSTOMER_ID")["TX_AMOUNT_DAILY"]
        .rolling(window=window, min_periods=1).std().reset_index(0, drop=True))
daily_spending["VAR_TX_AMOUNT_1D"] = daily_spending.groupby("CUSTOMER_ID")["TX_AMOUNT_DAILY"].diff()
df = df.merge(daily_spending, on=["CUSTOMER_ID", "TX_DATETIME"], how="left")


# EVALUACIÓN DIARIA + DETECCIÓN DE DRIFT

unique_dates = sorted(df['TX_DATETIME'].unique())
recall_list_eval = []
drifts_detectados = []
en_periodo_refractario = False
dias_desde_drift = 0
baseline_actual = np.nan

for i in range(0, len(unique_dates) - VENTANA):
    contador_inicial, contador_final = i, VENTANA + i
    df_p = df[df['TX_DATETIME'].isin(unique_dates[contador_inicial:contador_final])].copy()
    df_train = df_p[df_p['TX_DATETIME'].isin(unique_dates[contador_inicial:contador_final - 1])]
    df_prueba = df_p[df_p['TX_DATETIME'] == unique_dates[contador_final - 1]]

    try:
        df_ent = df_train[df_train['TX_FRAUD'] == 0].sample(
            n=df_train[df_train['TX_FRAUD'] == 1].shape[0], random_state=RUN_SEED)
    except ValueError:
        continue
    df_ent = pd.concat([df_ent, df_train[df_train['TX_FRAUD'] == 1]])

    features_to_drop = ['TX_DATETIME', 'TX_FRAUD_SCENARIO', 'TX_FRAUD',
                         'TRANSACTION_ID', 'TX_DATE', 'TX_TIME_SECONDS', 'TX_TIME_DAYS']
    columns_to_keep = [c for c in df_ent.columns if c not in features_to_drop]

    X_train, y_train = df_ent[columns_to_keep], df_ent['TX_FRAUD']
    X_test, y_test = df_prueba[columns_to_keep], df_prueba['TX_FRAUD']

    model = XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.1,
                           objective='binary:logistic', random_state=RUN_SEED, eval_metric='logloss')
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    recall_eval = recall_score(y_test, y_pred)
    recall_list_eval.append(recall_eval)

    if en_periodo_refractario:
        dias_desde_drift += 1
        if recall_eval >= baseline_actual or dias_desde_drift >= DIAS_REFRACTARIO:
            en_periodo_refractario = False
            dias_desde_drift = 0

    recall_suavizado = (np.median(recall_list_eval[-VENTANA_SUAVIZADO:])
                         if len(recall_list_eval) >= VENTANA_SUAVIZADO else recall_eval)

    if len(recall_list_eval) > 1 and not en_periodo_refractario:
        baseline_actual = np.median(recall_list_eval[:-1])
        if baseline_actual > 0:
            caida_relativa = (baseline_actual - recall_suavizado) / baseline_actual
            if caida_relativa > UMBRAL_RELATIVO:
                drifts_detectados.append({
                    'fecha': unique_dates[contador_final - 1],
                    'indice_local': len(recall_list_eval) - 1,
                    'caida_relativa': caida_relativa,
                })
                en_periodo_refractario = True
                dias_desde_drift = 0

print(f"Drifts detectados en {RUN} (ventana={VENTANA}): {len(drifts_detectados)}")
for d in drifts_detectados:
    print(f"  {pd.Timestamp(d['fecha']).date()}  caida_relativa={d['caida_relativa']:.2f}")


# ETIQUETADO

def etiqueta_corte(fecha):
    fecha = pd.Timestamp(fecha)
    return min(FECHAS_CORTE.items(), key=lambda kv: abs((fecha - kv[1]).days))[0]

for d in drifts_detectados:
    d['etiqueta'] = etiqueta_corte(d['fecha'])


# Serie temporal de recall con drifts marcados

baseline_recall = float(np.median(recall_list_eval))

fig, ax = plt.subplots(figsize=(14, 6))
ax.plot(range(len(recall_list_eval)), recall_list_eval,
        color=COLOR_RECALL, linewidth=1.8, label='Recall diario')
ax.axhline(baseline_recall, color=COLOR_BASELINE, linestyle='--', linewidth=1.2,
           label=f'Recall basal (mediana = {baseline_recall:.2f})')

etiquetas_usadas = set()
for d in drifts_detectados:
    x = d['indice_local']
    ax.axvline(x, color=COLOR_DRIFT, linestyle=':', linewidth=1.5, alpha=0.85,
               label='Drift detectado' if 'Drift detectado' not in etiquetas_usadas else None)
    etiquetas_usadas.add('Drift detectado')
    ax.annotate(d['etiqueta'], xy=(x, 1.0), xytext=(x, 1.03),
                ha='center', fontsize=10, color=COLOR_DRIFT, fontweight='bold',
                annotation_clip=False)

ax.set_ylim(0, 1.08)
ax.set_xlabel('Día de evaluación', fontsize=12, fontweight='bold')
ax.set_ylabel('Recall', fontsize=12, fontweight='bold')
ax.set_title(f'Serie temporal de recall — {RUN}, ventana = {VENTANA} días', fontsize=13)
ax.spines[['top', 'right']].set_visible(False)
ax.legend(loc='lower right', frameon=True)
plt.tight_layout()

plt.savefig('fig10_serie_recall.pdf', dpi=300, bbox_inches='tight')
plt.savefig('fig10_serie_recall.png', dpi=300, bbox_inches='tight')
plt.show()
```
